```python
# الخلية 2: فحص الصور والتحقق من سلامتها
# تقوم هذه الخلية بالبحث عن جميع الصور في المجلد، وفحص ما إذا كانت هناك صور تالفة، واستخراج الأحجام الأكثر شيوعاً.

import os
from PIL import Image
from collections import Counter

dataset_dir = 'diabetic_foot_ulcer'
image_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.gif')

image_paths = []
corrupted_images = []
image_sizes = []

for root, dirs, files in os.walk(dataset_dir):
    for file in files:
        if file.lower().endswith(image_extensions):
            full_path = os.path.join(root, file)
            image_paths.append(full_path)
            try:
                with Image.open(full_path) as img:
                    image_sizes.append(img.size)
            except Exception as e:
                corrupted_images.append((full_path, str(e)))

print(f"إجمالي عدد الصور المكتشفة: {len(image_paths)}")
print(f"عدد الصور التالفة (التي تعذر فتحها): {len(corrupted_images)}")

if corrupted_images:
    print("\nالملفات التالفة التي تحتاج إلى تنظيف:")
    for path, err in corrupted_images[:10]:
        print(f"- {path}: {err}")
else:
    print("\nلا توجد صور تالفة بشكل مباشر (جميع الصور تفتح بنجاح).")

if image_sizes:
    size_counts = Counter(image_sizes)
    print("\nأحجام الصور الأكثر شيوعاً (العرض × الارتفاع):")
    for size, count in size_counts.most_common(10):
        print(f"- {size}: {count} صورة")

    if len(size_counts) > 1:
        print("\nملاحظة: البيانات تحتوي على أحجام صور مختلفة. ستحتاج إلى توحيد الأحجام (Resize) عند تدريب النموذج.")
    else:
        print("\nملاحظة: جميع الصور لها نفس الحجم.")
```

```python
# الخلية 3: توحيد أحجام الصور (Image Resizing)
# تقوم هذه الخلية بتغيير حجم جميع الصور المكتشفة إلى حجم موحد (224x224) وتحويل صيغ الألوان إلى RGB لتهيئتها للنموذج.

import os
from PIL import Image

source_dir = 'diabetic_foot_ulcer'
target_dir = 'resized_images'
target_size = (224, 224)

image_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.gif')
resized_count = 0

if not os.path.exists(target_dir):
    os.makedirs(target_dir)

for root, dirs, files in os.walk(source_dir):
    for file in files:
        if file.lower().endswith(image_extensions):
            source_path = os.path.join(root, file)
            relative_path = os.path.relpath(root, source_dir)
            dest_folder = os.path.join(target_dir, relative_path)
            if not os.path.exists(dest_folder):
                os.makedirs(dest_folder)
            dest_path = os.path.join(dest_folder, file)
            try:
                with Image.open(source_path) as img:
                    if img.mode != 'RGB':
                        img = img.convert('RGB')
                    resized_img = img.resize(target_size, Image.Resampling.LANCZOS)
                    resized_img.save(dest_path)
                    resized_count += 1
            except Exception as e:
                print(f"فشل تعديل الصورة {file}: {e}")

print(f"تم الانتهاء بنجاح! تم توحيد حجم {resized_count} صورة وحفظها في المجلد '{target_dir}'.")
```

```python
# الخلية 4: استخراج ميزات البكسلات المسطحة (Feature Extraction - Flat Pixels)
# تحول الصور المعدلة إلى مصفوفات NumPy مسطحة ذات بعد واحد (1D) لتمثيل ميزات البكسلات الخام لكل صورة مع استخلاص التسميات (Labels).

import os
import numpy as np
from PIL import Image

target_dir = 'resized_images'
image_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.gif')

X = []
y = []

for root, dirs, files in os.walk(target_dir):
    for file in files:
        if file.lower().endswith(image_extensions):
            file_path = os.path.join(root, file)
            label = os.path.basename(root)
            try:
                with Image.open(file_path) as img:
                    if img.mode != 'RGB':
                        img = img.convert('RGB')
                    img_array = np.array(img)
                    flat_img = img_array.flatten()
                    X.append(flat_img)
                    y.append(label)
            except Exception as e:
                print(f"فشل في معالجة الملف {file_path}: {e}")

X = np.array(X)
y = np.array(y)

print(f"أبعاد مصفوفة الخصائص (X): {X.shape}")
print(f"أبعاد مصفوفة التسميات (y): {y.shape}")
print(f"عدد الميزات لكل صورة: {X.shape[1]}")
print(f"الفئات الفريدة المكتشفة: {np.unique(y)}")
```

```python
# الخلية 5: تصفية وتقسيم البيانات (Data Filtering & Splitting)
# تقوم بتصفية البيانات للاحتفاظ بالفئات الطبية الفعلية فقط [Abnormal, Normal] وتقسيمها إلى مجموعات تدريب واختبار مع تطبيع ميزات البكسل.

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# تصفية الفئات الطبية
target_classes = ['Abnormal(Ulcer)', 'Normal(Healthy skin)']
mask = np.isin(y, target_classes)
X_filtered = X[mask]
y_filtered = y[mask]

print(f"عدد العينات بعد التصفية: {X_filtered.shape[0]}")
print(f"توزيع الفئات المصفاة: {dict(zip(*np.unique(y_filtered, return_counts=True)))}")

# تقسيم البيانات
X_train, X_test, y_train, y_test = train_test_split(
    X_filtered, y_filtered, test_size=0.2, random_state=42, stratify=y_filtered
)

# تطبيع قيم البكسل
X_train_scaled = X_train / 255.0
X_test_scaled = X_test / 255.0
```

```python
# الخلية 6: تدريب وتقييم نموذج الغابة العشوائية (Random Forest Classifier)
# تدرب نموذج غابة عشوائية كلاسيكي على البيانات الخام المسطحة والمطبعة وتقيس مستوى دقته المبدئية.

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

print("جاري تدريب نموذج Random Forest...")
rf_model = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf_model.fit(X_train_scaled, y_train)
y_pred_rf = rf_model.predict(X_test_scaled)

print("\n--- تقرير أداء نموذج Random Forest ---")
print(f"الدقة الكلية (Accuracy): {accuracy_score(y_test, y_pred_rf):.2%}")
print(classification_report(y_test, y_pred_rf))
```

```python
# الخلية 7: تدريب وتقييم نموذج الآلات المتجهة الداعمة (SVM)
# تدرب نموذج SVM مع نواة خطية على ميزات البكسل وتقارن أدائه بنموذج الغابة العشوائية.

from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score

print("جاري تدريب نموذج SVM...")
svm_model = SVC(kernel='linear', random_state=42)
svm_model.fit(X_train_scaled, y_train)
y_pred_svm = svm_model.predict(X_test_scaled)

print("\n--- تقرير أداء نموذج SVM ---")
print(f"الدقة الكلية (Accuracy): {accuracy_score(y_test, y_pred_svm):.2%}")
print(classification_report(y_test, y_pred_svm))
```

```python
# الخلية 8: تثبيت مكتبات التصنيف المتقدمة
# تقوم بتثبيت مكتبات النماذج المعززة المتدرجة (XGBoost & LightGBM) المطلوبة للتصنيف الأكثر دقة.

!pip install -q xgboost lightgbm
```

```python
# الخلية 9: استخراج ميزات الحواف (HOG Feature Extraction)
# تستخلص ميزات مصفوفات الحواف والتدرج الموجه (HOG) من الصور بعد تحويلها لتدرج الرمادي لتقليل الميزات والتركيز على الهيكل الحوافي للقرح.

from skimage.feature import hog
from skimage.color import rgb2gray
import numpy as np

print("جاري استخراج ميزات الحواف باستخدام HOG...")
X_filtered_images = X_filtered.reshape(-1, 224, 224, 3)
X_hog = []

for img in X_filtered_images:
    gray_img = rgb2gray(img)
    features = hog(gray_img, orientations=8, pixels_per_cell=(16, 16),
                   cells_per_block=(1, 1), visualize=False)
    X_hog.append(features)

X_hog = np.array(X_hog)
print(f"أبعاد مصفوفة ميزات HOG الجديدة: {X_hog.shape}")
```

```python
# الخلية 10: تدريب نموذج XGBoost بميزات الحواف (HOG)
# تشفر التسميات الطبية لترميز رقمي (0 و 1)، وتقسم ميزات HOG، ثم تدرب نموذج XGBoost المتقدم وتقيس أدائه.

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import xgboost as xgb
from sklearn.metrics import classification_report, accuracy_score

le_final = LabelEncoder()
y_encoded = le_final.fit_transform(y_filtered)

X_train_hog, X_test_hog, y_train_hog, y_test_hog = train_test_split(
    X_hog, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

print("جاري تدريب نموذج XGBoost...")
xgb_model = xgb.XGBClassifier(n_estimators=500, max_depth=5, learning_rate=0.01, random_state=42, n_jobs=-1)
xgb_model.fit(X_train_hog, y_train_hog)
y_pred_xgb = xgb_model.predict(X_test_hog)

print("\n--- تقرير أداء نموذج XGBoost باستخدام ميزات HOG ---")
print(f"الدقة الكلية (Accuracy): {accuracy_score(y_test_hog, y_pred_xgb):.2%}")
print(classification_report(y_test_hog, y_pred_xgb, target_names=le_final.classes_))
```

```python
# الخلية 11: تدريب نموذج LightGBM بميزات الحواف (HOG)
# تدرب نموذج LightGBM السريع والفعال باستخدام ميزات HOG فقط وتقارن دقته التنبؤية بالنماذج السابقة.

import lightgbm as lgb
from sklearn.metrics import classification_report, accuracy_score

print("جاري تدريب نموذج LightGBM...")
lgb_model = lgb.LGBMClassifier(n_estimators=500, max_depth=5, learning_rate=0.01, random_state=42, n_jobs=-1, verbose=-1)
lgb_model.fit(X_train_hog, y_train_hog)
y_pred_lgb = lgb_model.predict(X_test_hog)

print("\n--- تقرير أداء نموذج LightGBM باستخدام ميزات HOG ---")
print(f"الدقة الكلية (Accuracy): {accuracy_score(y_test_hog, y_pred_lgb):.2%}")
print(classification_report(y_test_hog, y_pred_lgb, target_names=le_final.classes_))
```

```python
# الخلية 12: استخراج الميزات المدمجة (HOG + Color Histograms)
# تستخرج ميزات مدمجة تحتوي على هيكل الحواف ومدرجات ألوان القنوات الثلاثة (R, G, B) معاً لتحسين قدرة النموذج على تصنيف القرح بدقة عالية.

import cv2
import numpy as np
import pandas as pd
from skimage.feature import hog
from skimage.color import rgb2gray

print("جاري استخراج ميزات مدمجة (الحواف HOG + الألوان Color Histograms)...")
X_filtered_images = X_filtered.reshape(-1, 224, 224, 3)
X_combined_list = []

for img in X_filtered_images:
    gray_img = rgb2gray(img)
    hog_feat = hog(gray_img, orientations=8, pixels_per_cell=(16, 16),
                   cells_per_block=(1, 1), visualize=False)
    hist_r, _ = np.histogram(img[:, :, 0], bins=8, range=(0, 256), density=True)
    hist_g, _ = np.histogram(img[:, :, 1], bins=8, range=(0, 256), density=True)
    hist_b, _ = np.histogram(img[:, :, 2], bins=8, range=(0, 256), density=True)
    color_feat = np.concatenate([hist_r, hist_g, hist_b])
    combined_feat = np.concatenate([hog_feat, color_feat])
    X_combined_list.append(combined_feat)

X_combined_array = np.array(X_combined_list)
feature_names = [f"feat_{i}" for i in range(X_combined_array.shape[1])]
X_combined_features = pd.DataFrame(X_combined_array, columns=feature_names)

print(f"أبعاد المصفوفة المدمجة الجديدة (DataFrame): {X_combined_features.shape}")
```

```python
# الخلية 13: تقسيم البيانات المدمجة قبل التنظيف
# تقسم البيانات المدمجة المجمعة حديثاً إلى مجموعات تدريب واختبار لدراسة تأثير دمج الخصائص قبل عملية التطهير وإزالة المكررات.

from sklearn.model_selection import train_test_split

X_train_comb, X_test_comb, y_train_comb, y_test_comb = train_test_split(
    X_combined_features, y_filtered, test_size=0.2, random_state=42, stratify=y_filtered
)
```

```python
# الخلية 14: تدريب نموذج LightGBM على البيانات المدمجة (مع وجود التكرار المسبب للتسريب)
# تدرب نموذج LightGBM على البيانات المدمجة الخام وتقيس مستوى دقة مرتفع للغاية ناتج بشكل غير مباشر عن وجود تكرار صور داخل المجموعتين.

import lightgbm as lgb
from sklearn.metrics import classification_report, accuracy_score

print("جاري تدريب نموذج LightGBM...")
lgb_model = lgb.LGBMClassifier(n_estimators=500, max_depth=5, learning_rate=0.01, random_state=42, n_jobs=-1, verbose=-1)
lgb_model.fit(X_train_comb, y_train_comb)
y_pred_comb = lgb_model.predict(X_test_comb)

print("\n--- تقرير أداء نموذج LightGBM باستخدام ميزات HOG والألوان ---")
print(f"الدقة الكلية (Accuracy): {accuracy_score(y_test_comb, y_pred_comb):.2%}")
print(classification_report(y_test_comb, y_pred_comb))
```

```python
# الخلية 15: تنظيف البيانات وإلغاء التكرار لمنع تسريب البيانات (Data Leakage Removal)
# تكشف وتزيل كافة الصفوف والصور المكررة تماماً لحل مشكلة تسريب البيانات، وتعيد تقسيم البيانات إلى مجموعات خالية من التكرار.

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

df_temp = X_combined_features.copy()
df_temp['target_label'] = y_filtered

print(f"عدد العينات الكلي قبل حذف التكرار: {df_temp.shape[0]}")
df_cleaned = df_temp.drop_duplicates(subset=feature_names, keep='first')

print(f"عدد العينات الكلي بعد حذف التكرار: {df_cleaned.shape[0]}")
print(f"تم التخلص من {df_temp.shape[0] - df_cleaned.shape[0]} عينة مكررة بنجاح!")

X_cleaned = df_cleaned[feature_names]
y_cleaned = df_cleaned['target_label'].values

X_train_clean, X_test_cleaned, y_train_clean, y_test_cleaned = train_test_split(
    X_cleaned, y_cleaned, test_size=0.2, random_state=42, stratify=y_cleaned
)

y_train_clean_encoded = le_final.transform(y_train_clean)
y_test_clean_encoded = le_final.transform(y_test_cleaned)

print(f"\nحجم بيانات التدريب الجديدة: {X_train_clean.shape[0]}")
print(f"حجم بيانات الاختبار الجديدة: {X_test_cleaned.shape[0]}")
```

```python
# الخلية 16: التحقق المباشر من خلو البيانات من التسريب (Leakage Check)
# تقارن رياضياً وعبر المجموعات ميزات عينات التدريب والاختبار لضمان عدم تداخل أي عينة مكررة والتحقق من سلامة البيانات بنسبة 100%.

train_set_check = set(tuple(row) for row in X_train_clean.values)
test_set_check = set(tuple(row) for row in X_test_cleaned.values)
overlap_check = train_set_check.intersection(test_set_check)

print(f"عدد العينات المتطابقة الآن بين التدريب والاختبار: {len(overlap_check)}")
if len(overlap_check) == 0:
    print("✅ ممتاز! تم حل مشكلة تسريب البيانات بنجاح تام.")
```

```python
# الخلية 17: تدريب وتقييم نموذج XGBoost المثالي بالتحقق المتقاطع (XGBoost Model with CV)
# تطبق تقييم التحقق المتقاطع (5-Fold Cross Validation) على بيانات التدريب النظيفة ثم تدرب وتقيم أداء XGBoost النهائي على بيانات الاختبار.

from sklearn.model_selection import StratifiedKFold, cross_val_score
import xgboost as xgb
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

print("=== نموذج XGBoost المثالي (بدون تسريب بيانات) ===")

xgb_model_cv = xgb.XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.03,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    eval_metric='logloss'
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores_xgb = cross_val_score(xgb_model_cv, X_train_clean, y_train_clean_encoded, cv=cv, scoring='accuracy')

print(f"أداء التحقق المتقاطع (5-Fold Cross-Validation Accuracy): {cv_scores_xgb.mean():.2%} ± {cv_scores_xgb.std():.2%}")

print("\nجاري التدريب والتقييم النهائي على بيانات الاختبار...")
xgb_model_cv.fit(X_train_clean, y_train_clean_encoded)
y_pred_xgb_final = xgb_model_cv.predict(X_test_cleaned)

print(f"الدقة الكلية النهائية (Test Accuracy): {accuracy_score(y_test_clean_encoded, y_pred_xgb_final):.2%}")
print(classification_report(y_test_clean_encoded, y_pred_xgb_final, target_names=le_final.classes_))

plt.figure(figsize=(5,4))
sns.heatmap(confusion_matrix(y_test_clean_encoded, y_pred_xgb_final), annot=True, fmt='d', cmap='Blues',
            xticklabels=le_final.classes_, yticklabels=le_final.classes_)
plt.title('Confusion Matrix - XGBoost (Clean Data)')
plt.ylabel('الحقيقة (Actual)')
plt.xlabel('التوقع (Predicted)')
plt.show()
```

```python
# الخلية 18: تدريب وتقييم نموذج LightGBM المثالي بالتحقق المتقاطع (LightGBM Model with CV)
# تطبق التحقق المتقاطع خماسي الطيات وتدرب نموذج LightGBM النهائي على البيانات المنظفة تماماً مع رسم مصفوفة الالتباس.

from sklearn.model_selection import StratifiedKFold, cross_val_score
import lightgbm as lgb
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

print("=== نموذج LightGBM المثالي (بدون تسريب بيانات) ===")

lgb_model_cv = lgb.LGBMClassifier(
    n_estimators=300,
    max_depth=8,
    learning_rate=0.03,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    verbose=-1
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores_lgb = cross_val_score(lgb_model_cv, X_train_clean, y_train_clean_encoded, cv=cv, scoring='accuracy')

print(f"أداء التحقق المتقاطع (5-Fold Cross-Validation Accuracy): {cv_scores_lgb.mean():.2%} ± {cv_scores_lgb.std():.2%}")

print("\nجاري التدريب والتقييم النهائي على بيانات الاختبار...")
lgb_model_cv.fit(X_train_clean, y_train_clean_encoded)
y_pred_lgb_final = lgb_model_cv.predict(X_test_cleaned)

print(f"الدقة الكلية النهائية (Test Accuracy): {accuracy_score(y_test_clean_encoded, y_pred_lgb_final):.2%}")
print(classification_report(y_test_clean_encoded, y_pred_lgb_final, target_names=le_final.classes_))

plt.figure(figsize=(5,4))
sns.heatmap(confusion_matrix(y_test_clean_encoded, y_pred_lgb_final), annot=True, fmt='d', cmap='Greens',
            xticklabels=le_final.classes_, yticklabels=le_final.classes_)
plt.title('Confusion Matrix - LightGBM (Clean Data)')
plt.ylabel('الحقيقة (Actual)')
plt.xlabel('التوقع (Predicted)')
plt.show()
```

```python
# الخلية 20: حفظ النموذج وتوليد خادم الـ API (FastAPI Deployment Server)
# تحفظ النموذج النهائي وجهاز الترميز في ملفات صلبة، وتنشئ ملف خادم البايثون 'app.py' المتكامل بالخطوات الكاملة لتنظيف ومعالجة الصور المرفوعة من تطبيق فلاتر.

import joblib
import os

print("=== جاري حفظ النموذج والمحولات ===")
joblib.dump(lgb_model_cv, 'best_lightgbm_model.pkl')
joblib.dump(le_final, 'label_encoder.pkl')

print("✅ تم حفظ الملفات بنجاح:")
print("- best_lightgbm_model.pkl (حجم النموذج)")
print("- label_encoder.pkl (ترميز الفئات)")

fastapi_code = """import os
from fastapi import FastAPI, File, UploadFile
from fastapi.responses import JSONResponse
import uvicorn
import numpy as np
import joblib
from PIL import Image
import io
from skimage.feature import hog
from skimage.color import rgb2gray

app = FastAPI(title="Diabetic Foot Ulcer Detection API")

MODEL_PATH = "best_lightgbm_model.pkl"
ENCODER_PATH = "label_encoder.pkl"

if os.path.exists(MODEL_PATH) and os.path.exists(ENCODER_PATH):
    model = joblib.load(MODEL_PATH)
    le = joblib.load(ENCODER_PATH)
    print("✅ تم تحميل النموذج والمحولات بنجاح!")
else:
    raise FileNotFoundError("تنبيه: لم يتم العثور على ملفات النموذج المحفوظة.")

def preprocess_image(image_bytes):
    img = Image.open(io.BytesIO(image_bytes))
    if img.mode != 'RGB':
        img = img.convert('RGB')
    img_resized = img.resize((224, 224), Image.Resampling.LANCZOS)
    img_array = np.array(img_resized)
    gray_img = rgb2gray(img_array)
    hog_feat = hog(gray_img, orientations=8, pixels_per_cell=(16, 16),
                   cells_per_block=(1, 1), visualize=False)
    hist_r, _ = np.histogram(img_array[:, :, 0], bins=8, range=(0, 256), density=True)
    hist_g, _ = np.histogram(img_array[:, :, 1], bins=8, range=(0, 256), density=True)
    hist_b, _ = np.histogram(img_array[:, :, 2], bins=8, range=(0, 256), density=True)
    color_feat = np.concatenate([hist_r, hist_g, hist_b])
    combined_feat = np.concatenate([hog_feat, color_feat]).reshape(1, -1)
    return combined_feat

@app.post("/predict")
async def predict(file: UploadFile = File(...)):
    try:
        contents = await file.read()
        features = preprocess_image(contents)
        prediction_idx = model.predict(features)[0]
        prediction_prob = model.predict_proba(features)[0]
        class_name = le.inverse_transform([prediction_idx])[0]
        confidence = float(max(prediction_prob))
        return JSONResponse({
            "status": "success",
            "class_name": class_name,
            "confidence": confidence,
            "probabilities": {
                le.classes_[0]: float(prediction_prob[0]),
                le.classes_[1]: float(prediction_prob[1])
            }
        })
    except Exception as e:
        return JSONResponse({
            "status": "error",
            "message": str(e)
        }, status_code=500)

if __name__ == "__main__":
    uvicorn.run(app, host="0.0.0.0", port=8000)
"""

with open('app.py', 'w', encoding='utf-8') as f:
    f.write(fastapi_code)

print("\n✅ تم إنشاء ملف خادم البايثون 'app.py' بنجاح!")
print("💡 يمكنك الآن تشغيل هذا الخادم باستخدام الأمر: uvicorn app:app --reload")
print("\n📱 في كود Flutter، كل ما عليك فعله هو إرسال طلب POST للصورة كالتالي:")
print("""
// كود دارت المبسط في فلاتر للإرسال:
var request = http.MultipartRequest('POST', Uri.parse('http://YOUR_SERVER_IP:8000/predict'));
request.files.add(await http.MultipartFile.fromPath('file', imageFile.path));
var response = await request.send();
""")
```
